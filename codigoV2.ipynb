```python
import pandas as pd, unicodedata

IN  = "/content/sample_data/NOTIFICAR_VN_2025.xlsx"
OUT = "CNPJ_NOTIFICAR_VN_pipe_with_header.txt"

def clean(s):
    if pd.isna(s):
        return ""
    s = str(s).replace("\u00a0", " ").strip()
    return s

# Localiza a aba de forma automática
xls = pd.ExcelFile(IN)
sheet_candidates = [s for s in xls.sheet_names if "NOT_VAF_25_CNAE" in s.strip().upper()]

if sheet_candidates:
    SHEET = sheet_candidates[0]
else:
    # Caso não exista exatamente esse padrão, usa a primeira aba do arquivo
    SHEET = xls.sheet_names[0]

df = pd.read_excel(IN, sheet_name=SHEET, dtype=str).applymap(clean)
df = df[~(df.eq("").all(axis=1))]  # remove linhas totalmente vazias

cols = [c.strip() for c in df.columns]
lines = ["|" + "|".join(cols) + "|"] + ["|" + "|".join(r) + "|" for r in df.astype(str).values.tolist()]

open(OUT, "w", encoding="utf-8-sig", newline="").write("\r\n".join(lines) + "\r\n")

print("Aba usada:", SHEET)
print("OK ->", OUT)
```
